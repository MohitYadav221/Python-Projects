# i phone sales data analysis in india 

```python
highest_rated = df.sort_values(by = ['Star Rating'], ascending = False)

highest_rated = highest_rated.head(10)
print(highest_rated['Product Name'])
                               
```

```python
print(highest_rated)
```

```python
iphones = highest_rated['Product Name'].value_counts()
label = iphones.index
counts = highest_rated['Number Of Ratings']
figure = px.bar(highest_rated, x= label, y = counts,
                title= 'Number of Ratings of highest Rated iphones')
figure.show()
```

```python
iphones = highest_rated['Product Name'].value_counts()
label = iphones.index
counts = highest_rated['Number Of Reviews']
figure = px.bar(highest_rated, x =label,
                y = counts,
                title = 'Number of Reviews of Highest Rated iphones')
figure.show()
```

```python
import plotly.express as px

figure = px.scatter(
    data_frame=df,
    x='Number Of Ratings',
    y='Sale Price',
    size='Discount Percentage',
    trendline='ols',
    title='Relationship between Sale Price and Number Of Ratings of iPhone'
)

figure.show()
```

```python
import plotly.express as px

figure = px.scatter(
    data_frame=df,
    x='Number Of Ratings',
    y='Discount Percentage',
    size='Sale Price',
    trendline='ols',
    title='Relationship between Discount Percentage and Number Of Ratings'
)

figure.show()
```

```python
most_expensive = df.loc[df['Sale Price'].idxmax()]
least_expensive = df.loc[df['Sale Price'].idxmin()]

print('Most Expensive Product :')
print(most_expensive)

print('\nLeast Expensive Product :')
print(least_expensive)
```
