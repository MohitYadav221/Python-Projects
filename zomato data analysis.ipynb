```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df = pd.read_csv("Zomato data .csv")
df
```

```python
df.head()
```

# let's convert the data type of the "rate column to float and remove the denominaor

```python
def handleRate(value):
    value = str(value).split('/')
    value = value[0];
    return float(value)

df['rate'] = df['rate'].apply(handleRate)
print(df.head())
                            
```

# summary of the data frame

```python
df.info()
```

```python
df.isnull().sum()
```

```python
sns.countplot(x = df['listed_in(type)'])
plt.xlabel('type of restaurant')
```

```python
grouped_data = df.groupby('listed_in(type)')['votes'].sum()

result = grouped_data.reset_index()

plt.plot(result['listed_in(type)'], result['votes'], c='green', marker='o')
plt.xlabel('Type of Restaurant', color='red', fontsize=20)
plt.ylabel('Votes', color='red', fontsize=20)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.hist(df['rate'],bins = 5)
plt.title('Rating Distribution')
plt.show()
```

```python
couple_data = df['approx_cost(for two people)']
sns.countplot(x = couple_data)
```

```python
plt.figure(figsize =(6,6))
sns.boxplot(x = 'online_order', y = 'rate', data = df)
```

```python
pivot_table = df.pivot_table(
    index='listed_in(type)',
    columns='online_order',
    values='name',
    aggfunc='count',
    fill_value=0
)
sns.heatmap(pivot_table, annot = True , cmap = 'YlGnBu', fmt ='d')
plt.title('Heatmap')
plt.xlabel('online order')
plt.ylabel('listed in type')
plt.show()
```
